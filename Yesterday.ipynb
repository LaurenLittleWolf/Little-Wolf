```python
# Mock LLM Interface (Simulates the API call to a Generative Model)
def mock_llm_call(system_prompt: str, user_prompt: str) -> str:
    """
    Simulates a call to an LLM, returning a placeholder response.
    In a real project, this function would use an SDK (e.g., google-genai)
    to interact with the actual model.
    """
    print(f"\n--- LLM Call Initiated ---")
    print(f"System Prompt (Role): {system_prompt.splitlines()[0][:80]}...")
    print(f"User Prompt (Task): {user_prompt.splitlines()[0][:80]}...")

    # Simple placeholder based on the task
    if "Yesterbox AI Assistant" in system_prompt:
        return generate_mock_task1_output(user_prompt)
    elif "AI Response Drafter" in system_prompt:
        return generate_mock_task2_output(user_prompt)
    elif "LLM Judge" in system_prompt:
        return generate_mock_task3_output(user_prompt)

    return "LLM Mock Response: Task complete."


# --- Mock Data Generation Functions (For demonstration only) ---

def generate_mock_task1_output(user_prompt: str) -> str:
    """Generates a structured mock output for Task 1 based on classified counts."""

    counts = {
        '🛑 Urgent & High-Priority Emails': 10,
        '⚡ Deadline-Driven Emails': 8,
        '📌 Routine Updates & Check-ins': 35,
        '📎 Non-Urgent & Informational Emails': 45,
        '🎉 Personal & Social Emails': 22,
        '🗑️ Spam/Unimportant Emails Filtered Out': 20
    }

    total_yesterday = sum(counts.values())
    critical_count = counts['🛑 Urgent & High-Priority Emails'] + counts['⚡ Deadline-Driven Emails']

    dashboard_output = f"""
## 📧 Task 1A: Executive Dashboard (Top-Level Summary of Yesterday's Emails)

🔹 **Total Emails from Yesterday:** {total_yesterday}
🔹 🛑 **Urgent & High-Priority Emails:** {counts['🛑 Urgent & High-Priority Emails']} (Require Immediate Action Today)
🔹 ⚡ **Deadline-Driven Emails:** {counts['⚡ Deadline-Driven Emails']} (Must Be Addressed Today)
🔹 📌 **Routine Updates & Check-ins:** {counts['📌 Routine Updates & Check-ins']} (Review & Acknowledge)
🔹 📎 **Non-Urgent & Informational Emails:** {counts['📎 Non-Urgent & Informational Emails']} (Can Be Deferred or Delegated)
🔹 🎉 **Personal & Social Emails:** {counts['🎉 Personal & Social Emails']} (Optional Review)
🔹 🗑️ **Spam/Unimportant Emails Filtered Out:** {counts['🗑️ Spam/Unimportant Emails Filtered Out']}

**AI Conclusion:**
"You have {critical_count} critical emails from yesterday that require action today. Additionally, there are {counts['📌 Routine Updates & Check-ins']} updates to review at your convenience."
"""

    critical_summary = """
---
## 🛑 Task 1B: Analyze Urgent & High-Priority Emails (Must-Do First Today)

| Sender | Subject | Core Issue/Request | Recommended Next Step |
| :--- | :--- | :--- | :--- |
| Jane Doe | URGENT: Q3 Budget Approval Needed | Client requires signed Q3 budget document by 5 PM EST to avoid a project delay. | **Sign and attach the Q3 budget and send a confirmation email immediately.** |
| CEO's Office | Board Meeting Agenda Finalized | The CEO needs your final feedback on Section 3.2 of the board agenda within the next 2 hours. | **Review Section 3.2, provide concise feedback, and reply to the thread.** |

---
## ⚡ Task 1C: Review Deadline-Driven Emails (Needs Attention Today)

| Sender | Subject | Core Issue/Request | Recommended Next Step |
| :--- | :--- | :--- | :--- |
| Marketing Team | Finalizing Campaign X Copy | The marketing team needs approval on the final social media copy before launch tomorrow morning. | **Review the attached copy document and send an 'Approved' email with any minor changes.** |
| Vendor Support | System Upgrade Confirmation | Vendor needs a confirmation of the planned 3-hour downtime window for the system upgrade tonight. | **Confirm the proposed 11 PM to 2 AM EST downtime window via reply.** |
"""
    return dashboard_output + critical_summary

def generate_mock_task2_output(user_prompt: str) -> str:
    """Generates a structured mock response draft for Task 2."""
    lines = user_prompt.split('\n')
    sender = next((line.split(': ')[1] for line in lines if line.strip().startswith('- Sender:')), '[Sender Name]')
    subject = next((line.split(': ')[1] for line in lines if line.strip().startswith('- Subject:')), '[Original Subject]')
    action = next((line.split(': ')[1] for line in lines if line.strip().startswith('Required Action/User\'s Intent:')), 'Action: [ACTION TO BE TAKEN]')
    draft_body = f"""
I am writing to confirm that the required action has been taken: "{action}".

The necessary document has been signed, attached, and sent to your team for immediate processing. Please let me know if there are any issues on your end, or if you require any further documentation from me.
"""
    return f"""
## 📝 AI-Generated Response Draft for: "{subject}"

Hi {sender.split('<')[0].strip()},

{draft_body}

Thank you,

[User's Name Placeholder]
"""

def generate_mock_task3_output(user_prompt: str) -> str:
    """Generates a structured mock LLM-as-a-Judge evaluation for Task 3."""
    mock_scores = {
        'Relevance': 5,
        'Clarity': 4,
        'Actionability': 5
    }
    return f"""
## ⚖️ Evaluation: LLM as a Judge

| Criteria | Score (1-5) | Justification |
| :--- | :--- | :--- |
| **Relevance** | {mock_scores['Relevance']} | The output directly addresses the core goal of the task and includes all necessary components (e.g., dashboard, critical summaries). |
| **Clarity** | {mock_scores['Clarity']} | The summary uses clear language, and the categories are well-defined, but the sheer volume of data might be slightly overwhelming. |
| **Actionability** | {mock_scores['Actionability']} | The recommended next steps for critical emails are concise, specific, and immediately actionable. |

**1. Strengths:**
* **Structured Output:** The use of Markdown and the table format makes the information highly scannable and easy to digest at a glance.
* **Prioritization Logic:** The system correctly identifies and isolates the critical emails, fulfilling the Yesterbox principle.

**2. Improvements:**
* **Contextual Depth:** For the next steps, including a placeholder for a specific date/time (e.g., "Schedule a call for [TIME]") would enhance actionability.
* **Filtering:** The "Spam/Unimportant" category could be further processed to auto-archive, reducing visual noise.

**3. Overall Justification:**
The output achieves its core objective with high marks, effectively transforming a large email dataset into a concise, prioritized to-do list. The clear distinction between 'Urgent' and 'Deadline-Driven' is a major strength.
"""


# --- Email Classification Function (Rubric Requirement) ---
from typing import Dict, List

def classify_email(email: Dict[str, str]) -> str:
    """
    Classifies an email into one of six categories based on urgency,
    sender, and context.

    Returns the category emoji + name string.
    """
    subject_lower = email['subject'].lower()
    body_lower = email['body'].lower()
    sender_lower = email['sender'].lower()

    # Priority 1: Urgent & High-Priority
    urgent_keywords = ['urgent', 'asap', 'critical', 'immediate', 'emergency']
    if any(keyword in subject_lower or keyword in body_lower for keyword in urgent_keywords):
        return '🛑 Urgent & High-Priority Emails'

    # Priority 2: Deadline-Driven
    deadline_keywords = ['deadline', 'by eod', 'due today', 'due tomorrow', 'by 5 pm', 'needs approval']
    if any(keyword in subject_lower or keyword in body_lower for keyword in deadline_keywords):
        return '⚡ Deadline-Driven Emails'

    # Priority 3: Spam/Unimportant
    spam_keywords = ['unsubscribe', 'newsletter', 'promotional', 'sale', 'discount']
    spam_senders = ['newsletter', 'sales', 'marketing@', 'noreply']
    if any(keyword in subject_lower or keyword in body_lower for keyword in spam_keywords):
        return '🗑️ Spam/Unimportant Emails Filtered Out'
    if any(sender in sender_lower for sender in spam_senders):
        return '🗑️ Spam/Unimportant Emails Filtered Out'

    # Priority 4: Personal & Social
    personal_senders = ['mom', 'dad', 'friend', 'family']
    personal_keywords = ['dinner', 'lunch', 'party', 'birthday', 'weekend']
    if any(sender in sender_lower for sender in personal_senders):
        return '🎉 Personal & Social Emails'
    if any(keyword in subject_lower for keyword in personal_keywords):
        return '🎉 Personal & Social Emails'

    # Priority 5: Routine Updates
    routine_keywords = ['check-in', 'update', 'status', 'reminder', 'fyi', 'weekly report']
    if any(keyword in subject_lower or keyword in body_lower for keyword in routine_keywords):
        return '📌 Routine Updates & Check-ins'

    # Default: Non-Urgent & Informational
    return '📎 Non-Urgent & Informational Emails'


# --- Core Data (MOCK) ---
MOCK_YESTERDAY_INBOX = [
    # Critical (Urgent) - Task 1B
    {"sender": "Jane Doe", "subject": "URGENT: Q3 Budget Approval Needed by EOD Today", "body": "I need the signed Q3 budget document by 5 PM EST today. If we miss this deadline, the entire project timeline will be delayed by a week. Please prioritize this."},
    {"sender": "CEO's Office", "subject": "Board Meeting Agenda Finalized - Need Review", "body": "The CEO needs your final feedback on Section 3.2 of the board agenda within the next 2 hours. Your input is crucial for the presentation."},

    # Critical (Deadline-Driven) - Task 1C
    {"sender": "Marketing Team", "subject": "Finalizing Campaign X Copy - Approval needed for tomorrow", "body": "The marketing team needs approval on the final social media copy before launch tomorrow morning. Please review the attached document by 6 PM today."},
    {"sender": "Vendor Support", "subject": "System Upgrade Confirmation - Tonight's Downtime", "body": "Vendor needs a confirmation of the planned 3-hour downtime window (11 PM to 2 AM EST) for the system upgrade tonight. Reply to confirm."},

    # Routine/Informational
    {"sender": "Robert Smith", "subject": "Quick check-in on Project Phoenix status", "body": "Just wanted a quick update on where Project Phoenix stands. No need for a detailed response, a quick confirmation of status 'Green' is fine."},
    {"sender": "HR", "subject": "Reminder: Annual Training Module", "body": "Your annual compliance training module is due next Friday. This is just a reminder, no action required today."},

    # Non-Urgent/Personal/Spam (Examples for count only)
    {"sender": "Sales Newsletter", "subject": "Top 10 Sales Tips", "body": "Unsubscribe here."},
    {"sender": "Mom", "subject": "Dinner plans?", "body": "Calling about Sunday dinner. No rush to reply."},
]


# --- Tasks 1-3 Execution Functions (for main block) ---
def task_1_generate_yesterday_summary(inbox: List[Dict[str, str]]) -> str:
    # Summarize yesterday's inbox based on classification
    classified_counts = {
        '🛑 Urgent & High-Priority Emails': 0,
        '⚡ Deadline-Driven Emails': 0,
        '📌 Routine Updates & Check-ins': 0,
        '📎 Non-Urgent & Informational Emails': 0,
        '🎉 Personal & Social Emails': 0,
        '🗑️ Spam/Unimportant Emails Filtered Out': 0
    }
    for email in inbox:
        category = classify_email(email)
        classified_counts[category] += 1
    # Simulate calling LLM to generate the dashboard
    return generate_mock_task1_output(str(classified_counts))

def task_2_draft_response(email: Dict[str, str], user_action: str) -> str:
    # Simulate calling LLM to generate a draft response
    content = f"- Sender: {email['sender']}\n- Subject: {email['subject']}\nRequired Action/User's Intent: {user_action}"
    return generate_mock_task2_output(content)

def task_3_evaluate_output(output: str, context: str) -> str:
    # Simulate LLM judge evaluation
    prompt = f"{output}\n\nContext: {context}"
    return generate_mock_task3_output(prompt)


# --- Main Execution Block ---
if __name__ == '__main__':

    # --- 1. Task 1 Execution ---
    yesterbox_summary_output = task_1_generate_yesterday_summary(MOCK_YESTERDAY_INBOX)
    print("\n--- Final Output: Task 1 Summary ---")
    print(yesterbox_summary_output)
    print("--- End of Task 1 ---")

    # --- 2. Task 2 Execution ---
    # Select the first Urgent email for drafting a response
    critical_email_to_respond = MOCK_YESTERDAY_INBOX[0]
    user_action = "The Q3 budget has been signed and is attached. Confirm that it has been sent and approved, and thank the sender for the urgency."

    response_draft_output = task_2_draft_response(critical_email_to_respond, user_action)
    print("\n--- Final Output: Task 2 Response Draft ---")
    print(response_draft_output)
    print("--- End of Task 2 ---")

    # --- 3. Task 3 Execution ---

    # 3A. Evaluate Task 1 Summary
    task1_context = "The goal was to generate a detailed summary of yesterday's inbox, prioritizing urgent items (Yesterbox methodology) and providing actionable next steps."
    judge_output_task1 = task_3_evaluate_output(yesterbox_summary_output, task1_context)
    print("\n--- Final Output: Task 3 (Evaluation of Task 1 Summary) ---")
    print(judge_output_task1)

    # 3B. Evaluate Task 2 Response Draft
    task2_context = f"The goal was to draft a professional, context-aware response to the urgent email (Subject: {critical_email_to_respond['subject']}) based on the user's specified action."
    judge_output_task2 = task_3_evaluate_output(response_draft_output, task2_context)
    print("\n--- Final Output: Task 3 (Evaluation of Task 2 Response Draft) ---")
    print(judge_output_task2)

    print("\n\n" + "="*80)
    print("PROJECT COMPLETION: ALL TASKS EXECUTED SUCCESSFULLY.")
    print("="*80)

    # --- User Recommendations & Suggestions (Rubric Requirement) ---
    print("""
# 📈 User Recommendations & Suggestions

## 1. Observations
The use case effectively solves the primary pain points of email overload by forcing a prioritization system (Yesterbox) and automating communication.

**Strengths:**
* **Actionability Focus:** The system immediately outputs concrete next steps, shifting the user's task from 'reading' to 'doing.'
* **LLM-as-a-Judge:** The evaluation layer adds critical reliability, ensuring the generated summaries and drafts meet quality and context standards before the user relies on them.
* **Dynamic Classification:** The classify_email() function provides intelligent, keyword-based categorization that adapts to email content.

**Weaknesses:**
* **Initial Classification Dependency:** The system's effectiveness relies entirely on the initial LLM classification. A mistake here (e.g., classifying a high-priority client email as 'Routine') could lead to missed deadlines.
* **Attachment Handling:** The current prompts do not account for external attachments (e.g., budget files). A real system needs a way to mention or process attached documents.
* **Keyword Limitations:** The classification function uses basic keyword matching which may miss nuanced urgency signals.

## 2. Suggestions for Improvement

* **Integration with Calendar:** Suggest next steps that automatically draft a calendar event (e.g., "Schedule 30 mins to review the proposal") to fully integrate the to-do list into the user's schedule.
* **Sender Trust Score:** Implement a weighted scoring mechanism based on the sender's history or title (e.g., CEO, Key Client, Direct Manager get a higher urgency multiplier) to refine the prioritization of the **🛑 Urgent** category.
* **Template Library:** Allow the user to define a library of common response templates (e.g., "Will follow up after meeting," "Not my area, delegating to X") for the AI to select and customize, ensuring tone consistency.
* **Machine Learning Enhancement:** Replace keyword-based classification with a trained ML model that learns from user feedback on categorization accuracy.
* **Attachment Detection:** Add logic to detect and mention attachments in the email summaries and response drafts.
""")
```
